### Exercise 35 – All-pairs distances: broadcasting vs the Gram trick

**All-pairs distances, the engineering way.**
(a) Using the first 300 rows of `Z`, compute the pairwise Euclidean distance matrix `D300` of shape `(300, 300)` **without a loop**, via 3-D broadcasting `(300,1,4) - (1,300,4)`.
(b) *Before running anything*, compute by hand (in a comment) how many **MB** the intermediate `(1900, 1900, 4)` float64 array would take if you did the same for all patients.
(c) Compute the full `(1900, 1900)` matrix `D` **without the 3-D intermediate**, using the identity ‖a−b‖² = ‖a‖² + ‖b‖² − 2·a·b (one matrix product). Clip tiny negatives to 0 before the square root.
(d) Check with `np.allclose(..., atol=1e-6)` that `D[:300, :300]` equals `D300` (why is a tolerance needed on the diagonal?), and print the index of the patient most similar to patient 0 (excluding patient 0).

````python
# Write your code here
# (a) Pairwise Euclidean distance for the first 300 rows using 3-D broadcasting
Z300 = Z[:300]
# Broadcasting shape: (300, 1, n_features) - (1, 300, n_features) -> (300, 300, n_features)
diff = Z300[:, np.newaxis, :] - Z300[np.newaxis, :, :]
D300 = np.sqrt(np.sum(diff**2, axis=-1))

# (b) Intermediate memory calculation:
# 1900 * 1900 * 4 elements * 8 bytes (float64) = 115,520,000 bytes which is 115.52 MB

# (c) Using the identidy we will calculate the matrix D
Z_sq = np.sum(Z**2, axis=1, keepdims=True)
D_sq = Z_sq + Z_sq.T - 2 * np.dot(Z, Z.T)
D = np.sqrt(np.clip(D_sq, 0, None)) # Clip limits the tiny negatives to 0 as requested

# (d) Check with np.allclose and find the most similar patient to patient 0 (excluding patient 0)
# (d) Checking with np.allclose that D[:300,:300] is equal to D300
is_close = np.allclose(D[:300, :300], D300, atol=1e-6)

# Why tolerance is needed on the diagonal: 
# Floating-point arithmetic inaccuracies can result in tiny negative values 
# (for example -1e-15) on the self-distance diagonal when evaluating a^2 + a^2 - 2a^2, 
# which need to be clipped or handled with a tolerance.

# Find the most similar patient to patient 0 (excluding patient 0 itself)
D_copy = D[0].copy()
D_copy[0] = np.inf  # Ignore distance to self
nearest_patient = np.argmin(D_copy)
nearest_distance = D_copy[nearest_patient]

print(f"allclose: {is_close} | D shape: {D.shape} | nearest to patient 0: {nearest_patient} | distance: {nearest_distance:.3f}")
````

<details><summary><b>Hint 1</b></summary>

`A[:, np.newaxis, :]` has shape `(300, 1, 4)`; `A[np.newaxis, :, :]` has shape `(1, 300, 4)`. Subtracting them broadcasts to `(300, 300, 4)`.

</details>
<details><summary><b>Hint 2</b></summary>

Memory: number of elements × 8 bytes (float64).

</details>
<details><summary><b>Hint 3</b></summary>

Row norms squared: `(Z**2).sum(axis=1)`; the cross term is `Z @ Z.T`. Shapes `(1900,1)` + `(1,1900)` − `(1900,1900)`.

</details>
<details><summary><b>Hint 4</b></summary>

Floating-point error can give −1e-15 on the diagonal; `np.clip(D2, 0, None)` before `sqrt`. After the sqrt that error becomes ~1e-7, larger than `allclose`'s default `atol=1e-8` – hence the tolerance.

</details>

### Exercise 36 – Bootstrap confidence interval

**Bootstrap.** (a) Estimate a 95% confidence interval for the mean age at diagnosis: with `rng = np.random.default_rng(42)`, resample the ages with replacement 2,000 times, computing all resample means **without a loop** (draw a `(2000, n)` matrix of indices at once), then take the 2.5 and 97.5 percentiles.
(b) Repeat with `default_rng(7)` and only 200 resamples. In a comment: which of the three numbers (the point estimate, the CI width, the CI position) should change between (a) and (b), which should not, and why?

````python
# Write your code here
ages = X_filled[:, 0]
n = len(ages)

# (a) First bootstrap: rng with seed 42, 2,000 resamples
rng_42 = np.random.default_rng(42)
indices_42 = rng_42.integers(0, n, size=(2000, n))
resampled_means_42 = np.mean(ages[indices_42], axis=1)
ci_42 = np.percentile(resampled_means_42, [2.5, 97.5])

# (b) Second bootstrap: rng with seed 7, 200 resamples
rng_7 = np.random.default_rng(7)
indices_7 = rng_7.integers(0, n, size=(200, n))
resampled_means_7 = np.mean(ages[indices_7], axis=1)
ci_7 = np.percentile(resampled_means_7, [2.5, 97.5])

# Print out results matching the expected format
print(f"mean age: {np.mean(ages):.2f}")
print(f"seed 42, B=2000: [{ci_42[0]:.2f} {ci_42[1]:.2f}]")
print(f"seed 7,  B=200:  [{ci_7[0]:.2f} {ci_7[1]:.2f}]")

# Which of the three numbers (point estimate, CI width, CI position) should change between (a) and (b), which should not, and why?
# - Point estimate (mean age): Shouldn't change, because it's calculated from the same original dataset.
# - CI width: should change, because changing the number of resamples (B) changes the standard error/precision of the bootstrap distribution, 
#   and changing the random seed draws a completely different set of random samples.
# - CI position: should change slightly, because a different random seed (and number of resamples) shifts the percentile boundaries of the bootstrap distribution.
````
